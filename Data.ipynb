## Getting Data

```python
from pathlib import Path
import nflreadpy as nfl

data_dir = Path("data")
data_dir.mkdir(parents=True, exist_ok=True)

draft = nfl.load_draft_picks(
    list(range(2000, 2022))
).to_pandas()

draft = draft[
    draft["position"].isin(["QB", "WR", "RB"])
].copy()

draft = draft.rename(columns={"season": "draft_year"})

stats = nfl.load_player_stats(
    list(range(2000, 2025)),
    summary_level="reg"
).to_pandas()

player_ids = draft["gsis_id"].dropna().unique()

stats = stats[
    stats["player_id"].isin(player_ids)
].copy()

draft.to_csv(data_dir / "draft_picks.csv", index=False)
stats.to_csv(data_dir / "player_season_stats.csv", index=False)

print(f"Saved to: {data_dir.resolve()}")
print(f"Draft picks: {len(draft):,} rows")
print(f"Player seasons: {len(stats):,} rows")
```
